```python
import numpy as np
import math
import matplotlib.pyplot as plt
from Scripts.beams import BeamType
from Scripts.beams import BeamSystem
```

```python
beam_system = BeamSystem("Cantilever-end load")
arm_beam = BeamType("annulus", [25, 20], "Aluminum7075-T6", "arm_beam")
beam_system.add_node(0,0,0,"origin")
beam_system.add_node(500,0,0,"point_1")
beam_system.add_beam(arm_beam,"origin","point_1",np.array([0,1,0]),"beam")
beam_system.add_boundary_condition(0,"x","origin")
beam_system.add_boundary_condition(0,"y","origin")
beam_system.add_boundary_condition(0,"z","origin")
beam_system.add_boundary_condition(0,"theta x","origin")
beam_system.add_boundary_condition(0,"theta y","origin")
beam_system.add_boundary_condition(0,"theta z","origin")
beam_system.add_force(0,0,-4,"point_1")
# beam_system.add_moment(0,400,0,"point_1")
beam_system.solve_FEA()
L = 500
I = (math.pi/64) * (50**4 - 40**4)
E = 71.7
F = 4
# kuu 
# Z1  Z2
# Z2  Z3
y = - (F * L ** 3) / (3*E*I)
m = F * L
dydx = (F*L**2)/(2*E*I) - (F*L**2)/(E*I)

print(beam_system.beams[0].transformation_matrix)
print(beam_system.kuu @ beam_system.du)

# print(beam_system.kuu)
print("Cantilever—end load")
print("calculated z: " + str(y) + " | FEA z: " + str(beam_system.z_displacements[1]))
print("calculated theta: " + str(dydx) + " | FEA reaction theta: " + str(beam_system.y_angles[1]))
print("calculated moment: " + str(m) + " | FEA moment: " + str(beam_system.y_moments[0]))
print("calculated reaction: " + str(F) + " | FEA reaction z: " + str(beam_system.z_forces[0]))
```

```python
beam_system = BeamSystem("Simple Support Center Load")
arm_beam = BeamType("annulus", [25, 20], "Aluminum7075-T6", "arm_beam")
beam_system.add_node(0,0,0,"origin")
beam_system.add_node(500,0,0,"point_1")
beam_system.add_node(1000,0,0,"point_2")
beam_system.add_beam(arm_beam,"origin","point_1",np.array([0,1,0]),"beam")
beam_system.add_beam(arm_beam,"point_1","point_2",np.array([0,1,0]),"beam")
beam_system.add_boundary_condition(0,"x","origin")
beam_system.add_boundary_condition(0,"y","origin")
beam_system.add_boundary_condition(0,"z","origin")
beam_system.add_boundary_condition(0,"theta x","origin")
beam_system.add_boundary_condition(0,"theta y","origin")
beam_system.add_boundary_condition(0,"x","point_1")
beam_system.add_boundary_condition(0,"z","point_1")
beam_system.add_boundary_condition(0,"x","point_2")
beam_system.add_boundary_condition(0,"y","point_2")
beam_system.add_boundary_condition(0,"z","point_2")
beam_system.add_boundary_condition(0,"theta x","point_2")
beam_system.add_boundary_condition(0,"theta y","point_2")
beam_system.add_force(0,-4,0,"point_1")
beam_system.solve_FEA()

L = 1000
I = (math.pi/64) * (50**4 - 40**4)
E = 71.7
F = 4

y = -(F * (L**3)) / (48 * E * I)
r = F/2
dydx = -(F * L ** 2) / (16 * E * I)

print("Simple Support Center Load")
print("calculated y: " + str(y) + " | FEA y: " + str(beam_system.y_displacements[1]))
print("calculated moment: " + str(0) + " | FEA moment: " + str(beam_system.z_moments[1]))
print("calculated theta center: " + str(0) + " | FEA theta center: " + str(beam_system.z_angles[1]))
print("calculated theta origin: " + str(dydx) + " | FEA theta origin: " + str(beam_system.z_angles[0]))
print("calculated theta edge: " + str(-dydx) + " | FEA theta edge: " + str(beam_system.z_angles[2]))
print("calculated reaction y: " + str(r) + " | FEA reaction y: " + str(beam_system.y_forces[0]))
```

```python
beam_system = BeamSystem("Cantilever-moment load")
arm_beam = BeamType("annulus", [25, 20], "Aluminum7075-T6", "arm_beam")
beam_system.add_node(0,0,0,"origin")
beam_system.add_node(500,0,0,"point_1")
beam_system.add_beam(arm_beam,"origin","point_1",np.array([0,1,0]),"beam")
beam_system.add_boundary_condition(0,"x","origin")
beam_system.add_boundary_condition(0,"y","origin")
beam_system.add_boundary_condition(0,"z","origin")
beam_system.add_boundary_condition(0,"theta x","origin")
beam_system.add_boundary_condition(0,"theta y","origin")
beam_system.add_boundary_condition(0,"theta z","origin")
beam_system.add_boundary_condition(0,"theta x","point_1")
beam_system.add_boundary_condition(0,"theta z","point_1")
beam_system.add_moment(0,4,0,"point_1")
beam_system.solve_FEA()

L = 500
I = (math.pi/64) * (50**4 - 40**4)
E = 71.7
M = 4

y = (M * L ** 2) / (2 * E * I)
dydx = M * L / (E * I)

# print(beam_system.kuu)
print("Cantilever—moment load")
print("calculated z: " + str(y) + " | FEA z: " + str(beam_system.z_displacements[1]))
print("calculated theta: " + str(dydx) + " | FEA theta: " + str(beam_system.y_angles[1]))
print("calculated reaction force z: " + str(0) + " | FEA reaction z: " + str(beam_system.z_forces[0]))
print("calculated reaction moment y: " + str(-M) + " | FEA reaction moment y: " + str(beam_system.y_moments[0]))
```

```python
beam_system = BeamSystem("Multidimensional Simple Support Center Load")
arm_beam = BeamType("annulus", [25, 20], "Aluminum7075-T6", "arm_beam")
beam_system.add_node(-1000,-1000,-1000,"point_0")
beam_system.add_node(0,0,0,"point_1")
beam_system.add_node(1000,1000,1000,"point_2")
beam_system.add_beam(arm_beam,"point_0","point_1",np.array([0,-1,1]),"beam")
beam_system.add_beam(arm_beam,"point_1","point_2",np.array([0,-1,1]),"beam")
beam_system.add_boundary_condition(0,"x","point_0")
beam_system.add_boundary_condition(0,"y","point_0")
beam_system.add_boundary_condition(0,"z","point_0")
beam_system.add_boundary_condition(0,"x","point_2")
beam_system.add_boundary_condition(0,"y","point_2")
beam_system.add_boundary_condition(0,"z","point_2")
beam_system.add_force(0,-1,1,"point_1")
beam_system.solve_FEA()

L = math.sqrt(3 * 2000 ** 2)
I = (math.pi/64) * (50**4 - 40**4)
E = 71.7
F = math.sqrt(2)
# kuu 
# Z1  Z2
# Z2  Z3
y = -(F * (L**3)) / (48 * E * I)
r = F/2
dydx = -(F * L ** 2) / (16 * E * I)


print("Multidimensional Simple Support Center Load")
print("FEA x: " + str(beam_system.x_displacements[1]))
print("FEA y: " + str(beam_system.y_displacements[1]))
print("FEA z: " + str(beam_system.z_displacements[1]))
print("FEA x theta: " + str(beam_system.x_angles[1]))
print("FEA y theta: " + str(beam_system.y_angles[1]))
print("FEA z theta: " + str(beam_system.z_angles[1]))
print("FEA x reaction: " + str(beam_system.x_forces[0]))
print("FEA y reaction: " + str(beam_system.y_forces[0]))
print("FEA z reaction: " + str(beam_system.z_forces[0]))

total_fea_disp = math.sqrt(beam_system.x_displacements[1] ** 2 + beam_system.y_displacements[1] ** 2 + beam_system.z_displacements[1] ** 2)
total_fea_reaction = math.sqrt(beam_system.x_forces[0] ** 2 + beam_system.y_forces[0] ** 2 + beam_system.z_forces[0] ** 2)

print("Calculated total displacement: " + str(y) + " | Total displacement FEA: " + str(total_fea_disp))
print("Calculated total reaction: " + str(r) + " | Total reaction FEA: " + str(total_fea_reaction))

print("Transformation matrix: " + str(np.round(beam_system.beams[0].transformation_matrix,3)))
```
