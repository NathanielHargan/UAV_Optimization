```python
import numpy as np

# Definitions
Heat_Required = 40 * 10 ** 8  # Define heat required

# Coefficients matrix
A = np.array([
    [1.0, 2, 1, 0, 1, 0, 0, 0, 0, 0, 0],  # Hydrocarbons
    [1, 1, 1, 0, 0, 1, 0, 0, 0, 0, 0],     # NOx
    [140, 80, 1, 0, 0, 0, 1, 0, 0, 0, 0],   # SOx
    [100, 90, 90, 0, 0, 0, 0, 1, 0, 0, 0],  # CO
    [6, 2, 1, -1, 0, 0, 0, 0, 1, 0, 0],     # Particulate
    [14000 * 2000, 19000 * 2000, 22000 * 2000, 0, 0, 0, 0, 0, 0, -1, 1]  # Heating Value
])

# Right-hand side (constraints)
b = np.array([100, 100, 2000, 20000, 80, Heat_Required])

# Objective function coefficients (cost vector)
c = np.array([60, 100, 160, 5, 0, 0, 0, 0, 0, 0, 0])

# Initial basic variable indices
basic = np.array([4, 5, 6, 7, 8, 10])

# Simplex iteration
while True:
    col_to_basic = np.argmin(c)  
    opt_check = True
    for bas_i in basic:
        if c[bas_i] < 0:  # If any cost is negative, optimality is not achieved
            opt_check = False

    if opt_check:
        break

    # Find the row to leave the basis (minimum ratio test)
    row_to_non_basic = 0
    min_ratio = np.inf
    for b_i, b_val in enumerate(b):
        if A[b_i][col_to_basic] > 0:
            ratio = b_val / A[b_i][col_to_basic]
            if ratio < min_ratio:
                row_to_non_basic = b_i
                min_ratio = ratio

    print(f"Pivoting: Row {row_to_non_basic}, Column {col_to_basic}")

    # Pivoting: Update A, b, and c
    pivot_element = A[row_to_non_basic][col_to_basic]
    A[row_to_non_basic] /= pivot_element
    b[row_to_non_basic] /= pivot_element

    for b_i, b_val in enumerate(b):
        if b_i != row_to_non_basic:
            row_mult = A[b_i][col_to_basic]
            A[b_i] -= row_mult * A[row_to_non_basic]
            b[b_i] -= row_mult * b[row_to_non_basic]

    c -= c[col_to_basic] * A[row_to_non_basic]

    # Update the basic variable indices
    basic[col_to_basic] = row_to_non_basic

    print(f"Updated basic indices: {basic}")
    print(f"Updated solution: {b}")
    print(f"Updated objective function coefficients: {c}")
    print("---------------")
```
