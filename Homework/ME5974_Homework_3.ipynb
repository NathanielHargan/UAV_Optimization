Setup

```python
import numpy as np

def f_1(a):
    return 7 * a ** 2 - 20 * a + 22

def f_2(x):
    return 6.983 * x[0] ** 2 + 12.415 * x[1] ** 2 - x[0]

def c_2(x):
    return np.array([6.983 * 2 * x[0] - 1, 2 * 12.415 * x[1]])
```

Problem 1a

Alternate Equal Interval Search

```python
Final_Interval_Length = 0.005
d = 0.05

q = 1
Step_count = 0

while not (f_1(q * d) < f_1((q + 1) * d) and f_1(q * d) < f_1((q - 1) * d)):
    q += 1
    Step_count += 1

a_e = (q - 1) * d
a_u = (q + 1) * d
Interval_Length = a_u - a_e

while Interval_Length > Final_Interval_Length:
    Interval_Length = a_u - a_e
    a_a = a_e + Interval_Length *(1/3)
    a_b = a_e + Interval_Length *(2/3)
    fa_e = f_1(a_e)
    fa_a = f_1(a_a)
    fa_b = f_1(a_b)
    fa_u = f_1(a_u)
    if fa_a < fa_b:
        a_u = a_b
    else:
        a_e = a_a
    Step_count += 1

print("Step count: ", Step_count)
print("a_e: ", a_e)
print("a_u: ",a_u)
```

Problem 1b

Golden Section Search

```python
Final_Interval_Length = 0.005
d = 0.05

q = 3
phi = 2 / (np.sqrt(5) - 1) # golden ratio
Step_count = 0

a1 = d
a2 = d + d * phi
a3 = d + d * phi + d * phi ** 2

while not (f_1(a2) < f_1(a1) and f_1(a2) < f_1(a3)):
    a1 = a2
    a2 = a3 
    a3 = a3 + d * phi ** q
    q += 1
    Step_count += 1
    
a_e = a1
a_u = a3
Interval_Length = a_u - a_e

while Interval_Length > Final_Interval_Length:
    Interval_Length = a_u - a_e
    a_a = a_e + Interval_Length *(1 - 1/phi)
    a_b = a_e + Interval_Length *(1/phi)
    fa_e = f_1(a_e)
    fa_a = f_1(a_a)
    fa_b = f_1(a_b)
    fa_u = f_1(a_u)
    if fa_a < fa_b:
        a_u = a_b
    else:
        a_e = a_a
    Step_count += 1 
    
print("Step count: ", Step_count)
print("a_e: ", a_e)
print("a_u: ",a_u)
```

Problem 2a

Steepest Descent method

```python
ig = np.array([2,1])
a = 0.01
e = 0.0001
k = 0 

x = ig
while np.linalg.norm(c_2(x)) > e:
    d = -c_2(x)
    x = x + a * d
    k += 1 

print("steps:", k)
print("x:", x)
print("f(x):", f_2(x))

print("\nH = ", np.array([[6.983 * 2, 0],[0 , 2 * 12.415]]))
print("Eigenvalues = ", np.linalg.eigvals(np.array([[6.983 * 2, 0],[0 , 2 * 12.415]])))
```

Problem 2b

Conjugate Gradient method

```python
ig = np.array([2,1])
a = 0.01
e = 0.0001
k = 0
d = 0
c = np.array([0,0])

x = ig
while np.linalg.norm(c_2(x)) > e:
    c_prev = c
    d_prev = d
    c = c_2(x)
    if k > 1:
        beta = (np.linalg.norm(c)/np.linalg.norm(c_prev)) ** 2
    else:
        beta = 0
    d = -c_2(x) + beta * d_prev
    x = x + a * d
    k += 1

print("steps:", k)
print("x:", x)
print("f(x):", f_2(x))
```

Problem 2c

Modified Newton method

```python
ig = np.array([2,1])

a = 0.25
e = 0.0001
k = 0
H_inv = np.linalg.inv(np.array([[6.983 * 2, 0],[0 , 2 * 12.415]]))
x = ig
while np.linalg.norm(c_2(x)) > e:
    x = x - a * H_inv @ c_2(x)
    k += 1

print("steps:", k)
print("x:", x)
print("f(x):", f_2(x))
```

Problem 2d

DFP method

```python
ig = np.array([2,1])

a = 0.25
e = 0.0001
k = 0
A = np.linalg.inv(np.array([[1, 0],[0 , 1]]))
x = ig
while np.linalg.norm(c_2(x)) > e:
    c = c_2(x)
    d = - A @ c
    d_x = a * d
    x = x + d_x
    c_next = c_2(x)
    y = c_next - c
    
    B = np.dot(d_x.reshape([2,1]), d_x.reshape([1,2])) / np.dot(d_x, y)
    z = A @ y
    C = -np.dot(z.reshape([2,1]), z.reshape([1,2])) / np.dot(y, z)
    A = A + B + C
    k += 1

print("steps:", k)
print("x:", x)
print("f(x):", f_2(x))
```

Problem 2e

BFGS method

```python
ig = np.array([2,1])

a = 0.25
e = 0.001
k = 0
H = np.linalg.inv(np.array([[1, 0],[0 , 1]]))
x = ig
while np.linalg.norm(c_2(x)) > e:
    c = c_2(x)
    H_inv = np.linalg.inv(H)
    d = - H_inv @ c
    d_x = a * d
    x = x + d_x
    c_next = c_2(x)
    y = c_next - c
    D = np.dot(y.reshape([2,1]), y.reshape([1,2])) / np.dot(y, d_x)
    z = A @ y
    E = np.dot(c.reshape([2,1]), c.reshape([1,2])) / np.dot(c, d)
    H = H + D + E
    k += 1

print("steps:", k)
print("x:", x)
print("f(x):", f_2(x))
```
