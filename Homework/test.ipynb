```python
import random
import numpy as np
rarities_on_belt = np.array([0,0,0,0,0])

for i in range(600000):
    rarity = 0
    alive = True
    while alive and rarity < 4:
        if random.random() > 0.2:
            if random.random() < 0.062*2:
                rarity += 1
                if random.random() < 0.062*2:
                    rarity += 1
                    if random.random() < 0.062*2:
                        rarity += 1
                        if random.random() < 0.062*2:
                            rarity += 1
        else:
            alive = False
    
    if rarity > 4:
        rarity = 4
    rarities_on_belt[:(rarity+1)] += 1 
print(rarities_on_belt/(15000))
            
        
        
```
