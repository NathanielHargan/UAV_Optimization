```python
import numpy as np
import matplotlib.pyplot as plt
from Scripts.mission_profile_v3 import mission_profile

m1 = mission_profile("m1",
                     50,
                     4,
                     0.2*9.8, # acc_y
                     10, 10)

m1.add_segment_acc_x("takeoff", 60, 10, 5)


list_t = np.arange(0, 7.5, 0.1)


x = np.zeros(len(list_t))
y = np.zeros(len(list_t))
v_x = np.zeros(len(list_t))
v_y = np.zeros(len(list_t))
a_x = np.zeros(len(list_t))
a_y = np.zeros(len(list_t))

for i, t in enumerate(list_t):
    x[i], y[i], v_x[i], v_y[i], a_x[i], a_y[i], name = m1.time_solve(t)

fig, axs = plt.subplots(2, 3)

axs[0,0].set_title("t vs y")
axs[0,0].plot(list_t, y)
axs[0,0].grid()

axs[0,1].set_title("t vs v_y")
axs[0,1].plot(list_t, v_y)
axs[0,1].grid()


axs[0,2].set_title("t vs a_y")
axs[0,2].plot(list_t, a_y)
axs[0,2].grid()


axs[1,0].set_title("t vs x")
axs[1,0].plot(list_t, x)
axs[1,0].grid()

axs[1,1].set_title("t vs v_x")
axs[1,1].plot(list_t, v_x)
axs[1,1].grid()


axs[1,2].set_title("t vs a_x")
axs[1,2].plot(list_t, a_x)
axs[1,2].grid()

print(m1.t_values)
print(m1.y_coords)
print(m1.vel_y_values)
print(m1.acc_y_values)
```
