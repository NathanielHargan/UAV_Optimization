```python
import matplotlib.pyplot as plt
from Scripts.drone import Drone

def plot_drone_geometry(drone_geo):
    plt.scatter(drone_geo.nodes[:,0], drone_geo.nodes[:,1])
    for beam in drone_geo.beams:
        node_0_x = beam.start_node_coords[0]
        node_1_x = beam.end_node_coords[0]
        node_0_y = beam.start_node_coords[1]
        node_1_y = beam.end_node_coords[1]
        plt.plot([node_0_x,node_1_x],[node_0_y,node_1_y])

d1 = Drone("drone_test")
d1.create_drone(0.05,0.5,0.2,8)  # hub radius | nominal radius | number of blades
plot_drone_geometry(d1.beam_system)
```
