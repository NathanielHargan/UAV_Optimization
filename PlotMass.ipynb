```python
from Scripts.Mass_Analysis.drone_mass import DroneMass
from Scripts.Mass_Analysis.drone_mass import Battery
from Scripts.drone import Drone
from Scripts.Finite_Elements.Beam_FEA_System.beam_type import BeamType
import numpy as np
import matplotlib.pyplot as plt

battery_1 = Battery(4.292e-3, np.array([0, 86.75, -50]), np.array([260.5, 123.5, 63.5]))
battery_2 = Battery(4.292e-3, np.array([0, -86.75, -50]), np.array([260.5, 123.5, 63.5]))


# al: 0.098 lb/in^3
# 26.6 kN/m^3 =* 10^-9 *
arm_beam = BeamType("annulus", [29, 25], "Aluminum7075-T6", "arm_beam")
strut_beam = BeamType("annulus", [14, 12], "Aluminum7075-T6", "strut_beam")

d1 = Drone("drone_test", 500, 200, 8, arm_beam, strut_beam)

d1_mass = DroneMass(d1, [battery_1, battery_2])

plt.scatter(d1_mass.centroids[0],d1_mass.centroids[1])

for i in range(8):
    centroid = d1_mass.arm_beam_centroids[i]
    trans_vector =  d1_mass.arm_transformation[i][0] * d1_mass.drone.nominal_rad/2
    plt.plot([centroid[0],centroid[0]+trans_vector[0]],[centroid[1],centroid[1]+trans_vector[1]])

for i in range(8):
    centroid = d1_mass.strut_beam_centroids[i]
    trans_vector =  d1_mass.strut_transformation[i][0] * d1_mass.drone.strut_length/2
    plt.plot([centroid[0],centroid[0]+trans_vector[0]],[centroid[1],centroid[1]+trans_vector[1]])
```
